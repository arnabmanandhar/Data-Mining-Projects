```python
#Load dataset from csv file

import pandas as pd
dataset = pd.read_csv('voice.csv')
dataset.head()
```

```python
dataset.columns
```

```python
dataset.info()
```

```python
# convert label into numerical value using label encoding

from sklearn.preprocessing import LabelEncoder

labelencoder_y = LabelEncoder()
dataset['label'] = labelencoder_y.fit_transform(dataset['label'])
dataset.head()




```

```python
# store label in different variable

label = dataset['label']
dataset = dataset.drop('label', axis=1)
```

```python
dataset.head()
```

```python
label.head()
```

```python
dataset.describe()
```

```python
# convert the dataset into pandas dataframe

dataset = pd.DataFrame(dataset)
dataset.head()
```

```python
# normalize to the range 0 to 1

from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
dataset_scaled = scaler.fit_transform(dataset)

# Convert the scaled NumPy array back to a Pandas DataFrame
dataset = pd.DataFrame(dataset_scaled, columns=dataset.columns)
```

```python
dataset.head()
```

```python
dataset.describe()
```

```python
# split the dataset into train, test

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(dataset, label, test_size=0.2, random_state=42)
```

```python
# instantiate a knn model

from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=5)
```

```python
# train the model

knn.fit(X_train, y_train)

pred = knn.predict(X_test)
```

```python
pred
```

```python
print("ground truth: ",y_test.to_numpy())
print("predicted: ",pred)
```

```python
# create the confusion matrix

from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, pred)
cm
```

```python
# visualize the confusion matrix using seaborn

import seaborn as sns

sns.heatmap(cm, annot=True, fmt='d')
```

```python
# classification report

from sklearn.metrics import classification_report

print(classification_report(y_test, pred))
```
