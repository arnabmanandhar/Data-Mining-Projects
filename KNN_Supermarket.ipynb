 Loading the dataset

```python
import pandas as pd

df = pd.read_csv('super-market.csv')
df.head()
```

```python
df.info()
```

```python
df['Rating'].unique()
```

```python
# Define thresholds
def categorize_rating(rating):
    if rating < 5.0:
        return 'Poor'
    elif 5.0 <= rating < 7.0:
        return 'Average'
    elif 7.0 <= rating < 9.0:
        return 'Good'
    else:
        return 'Excellent'

df['Rating_Category'] = df['Rating'].apply(categorize_rating)
df[['Rating','Rating_Category']]
```

```python
df.drop('Rating', axis=1, inplace=True)
df.head()
```

```python
df.drop('Invoice ID', axis=1, inplace=True)
df.head()
```

```python
import pandas as pd


# Define time division function
def time_division(time):
    hour = int(time.split(':')[0])
    if 5 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Afternoon'
    elif 17 <= hour < 21:
        return 'Evening'
    else:
        return 'Night'

# Apply time division
df['Time_Category'] = df['Time'].apply(time_division)
df.head()
```

```python
df.drop('Time', axis=1, inplace=True)
df.head()
```

```python
df.drop('Date', axis=1, inplace=True)
df.head()
```

```python
df.info()
```

```python
# convert categorical columns into numerical columns

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

cat_columns = ['Branch', 'City', 'Product line', 'Payment', 'Customer type', 'Time_Category', 'Gender', 'Rating_Category']

for col in cat_columns:
    df[col] = le.fit_transform(df[col])
```

For K-Nearest Neighbors (KNN), Min-Max Scaling is generally preferred. Here’s why:

    Distance Calculation: KNN relies on distance calculations between data points. Min-Max Scaling ensures all features contribute equally to this distance measure, as it scales features to a uniform range, typically [0, 1].

    Uniform Scaling: Min-Max Scaling provides a consistent range for each feature, preventing features with larger ranges from disproportionately influencing the distance metric.

```python
rating = df['Rating_Category']
X = df.drop('Rating_Category', axis=1)
```

```python
# divide into train test split

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, rating, test_size=0.2, random_state=42)
```

```python
# Define distance metrics
distance_metrics = ['euclidean', 'manhattan', 'minkowski']
```

```python
# import  knn classfier

from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import f1_score
import matplotlib.pyplot as plt
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
y_pred = knn.predict(X_test)
```

```python
print("y-pred is: ", y_pred)
print(  "y-test is: ", y_test.values)
```

```python
# calculate the f1 score

f1_score(y_test, y_pred, average='weighted')
```

```python
# Initialize a dictionary to store results
results = {}

for metric in distance_metrics:
    # Initialize lists to store accuracy scores
    train_accuracy = []
    test_accuracy = []
    f1_scores = []

    # Perform KNN with different k values
    k_values = list(range(1, 31, 2))
    for k in k_values:
        knn = KNeighborsClassifier(n_neighbors=k, metric=metric)
        knn.fit(X_train, y_train)
        y_pred = knn.predict(X_test)

        train_accuracy.append(knn.score(X_train, y_train))
        test_accuracy.append(knn.score(X_test, y_test))
        f1_scores.append(f1_score(y_test, y_pred, average='weighted'))

    # Plot accuracy vs k values
    plt.figure(figsize=(10, 5))
    plt.plot(k_values, train_accuracy, label='Training Accuracy')
    plt.plot(k_values, test_accuracy, label='Testing Accuracy')
    plt.xlabel('k values')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.title(f'KNN with {metric} distance metric')
    plt.savefig(f'knn_accuracy_{metric}.png')
    plt.show()

    # Store results for comparison
    best_k = k_values[test_accuracy.index(max(test_accuracy))]
    results[metric] = {
        'best_k': best_k,
        'train_accuracy': max(train_accuracy),
        'test_accuracy': max(test_accuracy),
        'f1_score': max(f1_scores)
    }

    print(f'Best k value for {metric} metric: {best_k}')
    print(f'Training Accuracy: {max(train_accuracy)}')
    print(f'Testing Accuracy: {max(test_accuracy)}')
    print(f'F1 Score: {max(f1_scores)}')
```
